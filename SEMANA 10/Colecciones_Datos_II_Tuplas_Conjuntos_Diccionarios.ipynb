## 1. ¿Qué es una colección de datos?

Una **colección de datos** es una estructura que permite almacenar varios valores dentro de una misma variable.

Por ejemplo, si queremos guardar las notas de cinco estudiantes, podríamos crear cinco variables independientes:

```python
nota1 = 4.0
nota2 = 3.5
nota3 = 4.8
nota4 = 2.9
nota5 = 4.2
```

Esto funciona, pero rápidamente se vuelve incómodo. Una colección permite agrupar los datos:

```python
notas = [4.0, 3.5, 4.8, 2.9, 4.2]
```

En Python existen varias estructuras de colección. En esta clase estudiaremos:

| Estructura | Sintaxis típica | ¿Permite duplicados? | ¿Se puede modificar? | Idea principal |
|---|---|---:|---:|---|
| Tupla | `(a, b, c)` | Sí | No | Datos agrupados que no deben cambiar |
| Conjunto | `{a, b, c}` | No | Sí | Valores únicos y operaciones de conjuntos |
| Diccionario | `{"clave": valor}` | Las claves no | Sí | Relacionar una clave con un valor |

**Pregunta para discutir:** ¿por qué no usar siempre una sola estructura?

Porque cada estructura representa una idea diferente y ofrece operaciones diferentes.


## 2. Tuplas

Una **tupla** es una colección **ordenada e inmutable**.

- **Ordenada:** los elementos tienen una posición.
- **Inmutable:** una vez creada, no podemos cambiar, agregar o eliminar directamente sus elementos.
- Puede contener valores repetidos.
- Puede contener datos de diferentes tipos.

### Crear una tupla

La forma más habitual es utilizar paréntesis:

```python
coordenadas = (10, 20)
```

También podemos crear una tupla sin escribir los paréntesis explícitamente:

```python
colores = "rojo", "verde", "azul"
```

Para una clase introductoria conviene recomendar los paréntesis porque hacen más evidente la estructura.


````python
coordenadas = (10, 20)
print(coordenadas)
print(type(coordenadas))
````

### 2.1 Acceder a los elementos

Las tuplas utilizan **índices**, comenzando en `0`.

```text
          índice
           0      1      2
          ↓      ↓      ↓
frutas = ("manzana", "pera", "uva")
```

Por tanto:

```python
frutas[0]  # "manzana"
frutas[1]  # "pera"
frutas[2]  # "uva"
```

También existen índices negativos:

- `-1` → último elemento
- `-2` → penúltimo elemento


````python
frutas = ("manzana", "pera", "uva")

print(frutas[0])
print(frutas[1])
print(frutas[-1])
````

### 2.2 Recorrer una tupla

Como la tupla es iterable, podemos recorrerla con `for`.

```python
for fruta in frutas:
    print(fruta)
```

Esto conecta directamente con la clase anterior sobre ciclos.


````python
frutas = ("manzana", "pera", "uva")

for fruta in frutas:
    print("Fruta:", fruta)
````

### 2.3 Longitud y pertenencia

Podemos usar:

- `len(tupla)` para saber cuántos elementos contiene.
- `in` para comprobar si un elemento pertenece a la tupla.
- `not in` para comprobar que no pertenece.



````python
numeros = (10, 20, 30, 40)

print("Cantidad de elementos:", len(numeros))
print(20 in numeros)
print(100 in numeros)
print(50 not in numeros)
````

### 2.4 ¿Qué significa que una tupla sea inmutable?

Después de crear:

```python
punto = (3, 5)
```

no podemos hacer:

```python
punto[0] = 10
```

Python producirá un error porque la tupla no permite modificar sus elementos.

Esto es útil cuando queremos representar información que conceptualmente debe permanecer fija, por ejemplo:

- coordenadas `(x, y)`;
- dimensiones de una imagen `(ancho, alto)`;
- días de la semana;
- meses;
- configuraciones que no deberían cambiar accidentalmente.



````python
punto = (3, 5)

# La siguiente línea produciría TypeError:
# punto[0] = 10
````

### 2.5 Desempaquetado de tuplas

Podemos asignar los elementos de una tupla a varias variables:

```python
punto = (4, 7)
x, y = punto
```

Ahora:

```python
x == 4
y == 7
```

Esta operación se llama **desempaquetado** (*unpacking*).


````python
punto = (4, 7)

x, y = punto

print("x =", x)
print("y =", y)
````

### 2.6 Métodos útiles de las tuplas

Dos métodos especialmente importantes son:

- `count(valor)` → cuenta cuántas veces aparece un valor.
- `index(valor)` → devuelve la posición de la primera aparición.



````python
datos = (2, 5, 2, 8, 2, 10)

print("Cantidad de veces que aparece 2:", datos.count(2))
print("Primera posición de 8:", datos.index(8))
````

## 3. Conjuntos (`set`)

Un **conjunto** almacena elementos **únicos** y no se utiliza para representar una secuencia ordenada.

Una característica fundamental:

> Si agregamos varias veces el mismo valor a un conjunto, solo se conserva una copia.

### Crear un conjunto

```python
numeros = {1, 2, 3, 4}
```

Para crear un conjunto vacío debemos usar:

```python
conjunto_vacio = set()
```

**Importante:** `{}` crea un diccionario vacío, no un conjunto vacío.


````python
numeros = {1, 2, 3, 4, 4, 4}

print(numeros)
print(type(numeros))

conjunto_vacio = set()
print(type(conjunto_vacio))
````

### 3.1 ¿Por qué desaparecen los repetidos?

El conjunto representa una colección de **valores únicos**.

Esto lo hace muy útil para problemas como:

- encontrar estudiantes únicos;
- eliminar registros repetidos;
- conocer las categorías presentes en un conjunto de datos;
- comparar grupos de elementos.



````python
usuarios = {"Ana", "Luis", "Ana", "Carlos", "Luis"}

print(usuarios)
print("Cantidad de usuarios únicos:", len(usuarios))
````

### 3.2 Agregar y eliminar elementos

Métodos principales:

- `add(elemento)` → agrega un elemento.
- `remove(elemento)` → elimina un elemento; produce error si no existe.
- `discard(elemento)` → elimina un elemento si existe; si no existe, no produce error.
- `clear()` → elimina todos los elementos.



````python
frutas = {"manzana", "pera"}

frutas.add("uva")
print(frutas)

frutas.discard("pera")
print(frutas)

frutas.discard("banana")  # No genera error
print(frutas)
````

### 3.3 Operaciones entre conjuntos

Los conjuntos permiten realizar operaciones matemáticas conocidas:

#### Unión

Elementos que están en **A o en B**.

```python
A | B
```

#### Intersección

Elementos que están en **A y en B**.

```python
A & B
```

#### Diferencia

Elementos que están en A pero **no** en B.

```python
A - B
```

#### Diferencia simétrica

Elementos que están en A o B, pero **no en ambos**.

```python
A ^ B
```


````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Unión:", A | B)
print("Intersección:", A & B)
print("A - B:", A - B)
print("B - A:", B - A)
print("Diferencia simétrica:", A ^ B)
````

### 3.4 Subconjuntos

También podemos preguntar si todos los elementos de un conjunto están contenidos en otro.

```python
A.issubset(B)
```

Otra forma:

```python
A <= B
```

Y para saber si un conjunto contiene a otro:

```python
A.issuperset(B)
```

o:

```python
A >= B
```


````python
grupo_A = {1, 2}
grupo_B = {1, 2, 3, 4}

print(grupo_A.issubset(grupo_B))
print(grupo_B.issuperset(grupo_A))
````

## 4. Diccionarios

Un **diccionario** almacena información en forma de **pares clave–valor**.

La idea es diferente a la de una secuencia:

```text
clave  →  valor
nombre →  "Ana"
edad   →  20
curso  →  "Python"
```

Se escribe:

```python
estudiante = {
    "nombre": "Ana",
    "edad": 20,
    "curso": "Python"
}
```

Características importantes:

- Cada clave identifica un valor.
- Las claves deben ser únicas.
- El diccionario es modificable.
- Podemos agregar, modificar y eliminar pares.
- Los diccionarios son especialmente útiles para representar entidades o registros.


````python
estudiante = {
    "nombre": "Ana",
    "edad": 20,
    "curso": "Python"
}

print(estudiante)
print(type(estudiante))
````

### 4.1 Consultar valores mediante claves

Para acceder a un valor utilizamos su clave:

```python
estudiante["nombre"]
```

No pensamos en "posición 0" como en una tupla. Pensamos en **qué información necesitamos** y usamos la clave correspondiente.


````python
estudiante = {
    "nombre": "Ana",
    "edad": 20,
    "curso": "Python"
}

print("Nombre:", estudiante["nombre"])
print("Edad:", estudiante["edad"])
print("Curso:", estudiante["curso"])
````

### 4.2 Agregar y modificar información

Si asignamos un valor a una clave que no existe, se crea un nuevo elemento:

```python
estudiante["programa"] = "Ingeniería"
```

Si la clave ya existe, se modifica su valor:

```python
estudiante["edad"] = 21
```


````python
estudiante = {
    "nombre": "Ana",
    "edad": 20
}

estudiante["programa"] = "Ingeniería"
estudiante["edad"] = 21

print(estudiante)
````

### 4.3 Eliminar información

Podemos utilizar:

- `del diccionario[clave]`
- `pop(clave)` → elimina y devuelve el valor.
- `clear()` → elimina todos los elementos.



````python
estudiante = {
    "nombre": "Ana",
    "edad": 20,
    "curso": "Python"
}

edad = estudiante.pop("edad")

print("Edad eliminada:", edad)
print(estudiante)
````

### 4.4 `get()` para consultar claves de forma segura

Si hacemos:

```python
estudiante["correo"]
```

y la clave no existe, Python genera un `KeyError`.

Con `get()` podemos proporcionar un valor por defecto:

```python
estudiante.get("correo", "No registrado")
```

Esto es muy útil cuando trabajamos con datos que pueden estar incompletos.


````python
estudiante = {
    "nombre": "Ana",
    "edad": 20
}

print(estudiante.get("nombre"))
print(estudiante.get("correo"))
print(estudiante.get("correo", "No registrado"))
````

### 4.5 Recorrer un diccionario

Podemos recorrer:

#### Solo las claves

```python
for clave in estudiante:
    print(clave)
```

#### Claves explícitamente

```python
for clave in estudiante.keys():
    print(clave)
```

#### Valores

```python
for valor in estudiante.values():
    print(valor)
```

#### Claves y valores

La forma más útil suele ser:

```python
for clave, valor in estudiante.items():
    print(clave, valor)
```


````python
estudiante = {
    "nombre": "Ana",
    "edad": 20,
    "curso": "Python"
}

for clave, valor in estudiante.items():
    print(clave, "->", valor)
````

## 5. Comparación: tuplas, conjuntos y diccionarios

Antes de elegir una estructura, podemos hacernos tres preguntas:

### Pregunta 1: ¿Necesito conservar una secuencia y permitir repetidos?

→ **Tupla**

Ejemplo: coordenadas, dimensiones, datos fijos.

### Pregunta 2: ¿Necesito valores únicos y operaciones entre grupos?

→ **Conjunto**

Ejemplo: estudiantes inscritos en dos cursos, categorías únicas, eliminación de duplicados.

### Pregunta 3: ¿Necesito asociar una característica con un valor?

→ **Diccionario**

Ejemplo: información de un estudiante, producto-precio, país-capital.

| Problema | Estructura recomendada |
|---|---|
| `(x, y)` | Tupla |
| Valores únicos | Conjunto |
| `nombre → edad` | Diccionario |
| Comparar dos grupos | Conjunto |
| Registro con atributos | Diccionario |
| Datos fijos en una secuencia | Tupla |


## 6. Una misma situación, tres formas de pensar

Supongamos que tenemos las materias de un estudiante.

### Como tupla

```python
materias = ("Matemáticas", "Física", "Programación")
```

Representamos una secuencia que no pretendemos modificar.

### Como conjunto

```python
materias = {"Matemáticas", "Física", "Programación"}
```

Nos interesa principalmente tener materias únicas.

### Como diccionario

```python
materias = {
    "Matemáticas": 4.5,
    "Física": 3.8,
    "Programación": 4.7
}
```

Ahora cada materia está relacionada con una nota.

**Conclusión:** la información puede parecer similar, pero la estructura cambia según la pregunta que queramos responder.


````python
materias = {
    "Matemáticas": 4.5,
    "Física": 3.8,
    "Programación": 4.7
}

for materia, nota in materias.items():
    print(f"{materia}: {nota}")
````

## 7. Ejemplo guiado integrador: cursos y estudiantes

Tenemos dos cursos:

- Curso A: Ana, Luis, María, Carlos
- Curso B: Luis, María, Sofía, Juan

Queremos responder:

1. ¿Qué estudiantes están en ambos cursos?
2. ¿Qué estudiantes están solamente en el curso A?
3. ¿Cuántos estudiantes únicos existen?
4. Queremos almacenar además la nota de cada estudiante.

Primero usamos conjuntos para las preguntas de pertenencia y comparación.


````python
curso_A = {"Ana", "Luis", "María", "Carlos"}
curso_B = {"Luis", "María", "Sofía", "Juan"}

print("En ambos cursos:", curso_A & curso_B)
print("Solo en A:", curso_A - curso_B)
print("Todos los estudiantes:", curso_A | curso_B)
print("Cantidad de estudiantes únicos:", len(curso_A | curso_B))
````

Ahora usamos un diccionario para asociar cada estudiante con su nota.

```python
notas = {
    "Ana": 4.5,
    "Luis": 3.8,
    ...
}
```

Podemos recorrerlo y tomar decisiones con `if`.


````python
notas = {
    "Ana": 4.5,
    "Luis": 3.8,
    "María": 4.2,
    "Carlos": 2.9,
    "Sofía": 4.8,
    "Juan": 3.4
}

for estudiante, nota in notas.items():
    if nota >= 3.0:
        estado = "Aprobó"
    else:
        estado = "No aprobó"

    print(f"{estudiante}: {nota} - {estado}")
````

## 8. Mini-reto guiado para resolver en clase

### Situación

Una tienda tiene productos y precios:

```python
precios = {
    "pan": 2500,
    "leche": 4200,
    "huevos": 15000,
    "arroz": 6000
}
```

El programa debe:

1. Mostrar todos los productos.
2. Mostrar los precios.
3. Mostrar producto y precio juntos.
4. Preguntar al usuario por un producto.
5. Mostrar su precio si existe.
6. Si no existe, informar que el producto no está registrado.

### Antes de programar

Preguntar a los estudiantes:

- ¿Por qué un diccionario es adecuado?
- ¿Cuál sería la clave?
- ¿Cuál sería el valor?
- ¿Qué método podemos utilizar para evitar un `KeyError`?


````python
precios = {
    "pan": 2500,
    "leche": 4200,
    "huevos": 15000,
    "arroz": 6000
}

print("Productos:")
for producto in precios:
    print(producto)

print("\nPrecios:")
for precio in precios.values():
    print(precio)

print("\nProducto y precio:")
for producto, precio in precios.items():
    print(f"{producto}: ${precio}")

producto = input("\nIngrese un producto: ").lower()

precio = precios.get(producto)

if precio is not None:
    print(f"El precio de {producto} es ${precio}.")
else:
    print("El producto no está registrado.")
````

## 9. Errores frecuentes

### Error 1: confundir `{}` con un conjunto vacío

```python
vacio = {}
print(type(vacio))
```

Esto produce un **diccionario**, no un conjunto.

Para un conjunto vacío:

```python
vacio = set()
```

### Error 2: intentar modificar una tupla

```python
punto = (2, 3)
# punto[0] = 10
```

No es posible porque las tuplas son inmutables.

### Error 3: usar un índice para un diccionario

```python
estudiante = {"nombre": "Ana", "edad": 20}

# estudiante[0]  # No corresponde a la forma de acceso de un diccionario
```

En un diccionario normalmente accedemos mediante una **clave**:

```python
estudiante["nombre"]
```

### Error 4: olvidar que las claves de un diccionario son únicas

Si escribimos:

```python
datos = {"a": 10, "a": 20}
```

la segunda asignación reemplaza a la primera.

### Error 5: asumir que un conjunto mantiene un orden

Un conjunto debe pensarse como una colección de elementos únicos, no como una secuencia para acceder mediante índices.
