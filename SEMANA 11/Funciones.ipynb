## 1. ¿Qué es una función?

Una **función** es un bloque de código reutilizable que realiza una tarea específica.

Una forma útil de pensar en ella es:

**Entrada → Proceso → Salida**

No todas las funciones tienen entradas o salidas explícitas.

### ¿Por qué usar funciones?
- Evitan repetir código.
- Facilitan la lectura.
- Dividen problemas grandes en partes pequeñas.
- Facilitan pruebas y correcciones.
- Permiten reutilizar soluciones.
- Hacen más mantenibles los programas.


````python
def saludar():
    print("Hola, bienvenido al curso de Herramientas Computacionales.")

saludar()
saludar()
````

### Lectura de la sintaxis

```python
def saludar():
```

- `def` inicia una definición.
- `saludar` es el nombre.
- `()` contiene los parámetros.
- `:` inicia el bloque.
- La indentación indica qué instrucciones pertenecen a la función.
- `saludar()` es la llamada.

**Pregunta:** ¿qué ocurre si definimos la función pero nunca escribimos `saludar()`?

**Respuesta:** la función queda definida, pero su cuerpo no se ejecuta.


## 2. Parámetros y argumentos

```python
def saludar(nombre):
    print("Hola", nombre)
```

Aquí `nombre` es un **parámetro**.

En:

```python
saludar("Laura")
```

`"Laura"` es un **argumento**.

**Regla:** parámetro = aparece en la definición; argumento = valor entregado en la llamada.


````python
def presentar(nombre, edad):
    print("Mi nombre es", nombre)
    print("Tengo", edad, "años")

presentar("Carlos", 20)
````

### Parámetros con valores por defecto

```python
def potencia(base, exponente=2):
    return base ** exponente
```

Si no se proporciona `exponente`, Python utiliza `2`.

También podemos usar argumentos con nombre:

```python
potencia(base=2, exponente=4)
```


````python
def potencia(base, exponente=2):
    return base ** exponente

print(potencia(5))
print(potencia(5, 3))
print(potencia( exponente=4, base= 2))
````

## 3. Valores de retorno: `return`

`return` entrega un resultado al lugar desde donde se llamó la función y termina su ejecución en ese punto.

```python
def sumar(a, b):
    resultado = a + b
    return resultado
```

Después podemos hacer:

```python
total = sumar(4, 7)
```

### `print()` no es lo mismo que `return`

`print()` muestra información en pantalla.

`return` entrega información al programa, por lo que podemos almacenarla, combinarla o utilizarla en otra función.


````python
def sumar_print(a, b):
    print(a + b)

def sumar_return(a, b):
    return a + b

resultado_1 = sumar_print(4, 7)
print("resultado_1 =", resultado_1)

resultado_2 = sumar_return(4, 7)
print("resultado_2 =", resultado_2)

doble = 2 * sumar_return(4, 7)
print("Doble de la suma =", doble)
````

### Ejemplo guiado — Conversión de temperatura

$$F = \frac{9}{5}C + 32$$

Identificamos:
- Entrada: `celsius`.
- Proceso: aplicar la fórmula.
- Salida: Fahrenheit.

```python
def celsius_a_fahrenheit(celsius):
    return (9/5) * celsius + 32
```

**Preguntas:** ¿qué representa `celsius`? ¿Por qué usamos `return`? ¿Qué debería dar `0 °C`?

Respuesta: `32 °F`.


````python
def celsius_a_fahrenheit(celsius):
    return (9/5) * celsius + 32

print(celsius_a_fahrenheit(0))
print(celsius_a_fahrenheit(20))
print(celsius_a_fahrenheit(100))
````

## 4. Una función puede llamar a otra

Podemos construir soluciones grandes usando funciones pequeñas:

```python
def cuadrado(x):
    return x**2

def suma_de_cuadrados(a, b):
    return cuadrado(a) + cuadrado(b)
```

Esto reduce la repetición y permite reutilizar componentes.


````python
def cuadrado(x):
    return x**2

def suma_de_cuadrados(a, b):
    return cuadrado(a) + cuadrado(b)

print(suma_de_cuadrados(3, 4))
````

## 5. Alcance de variables (scope)

El **alcance** indica dónde puede utilizarse una variable.

### Variable local

```python
def calcular():
    resultado = 10
    print(resultado)
```

`resultado` es local a la función. Intentar usarla fuera produce un `NameError`.

### Variable global

```python
mensaje = "Hola"

def mostrar():
    print(mensaje)
```

Una función puede leer una variable definida fuera de ella.

**Buena práctica:** preferir parámetros y `return` antes que depender de muchas variables globales.


````python
mensaje = "Hola desde el programa"

def mostrar_mensaje():
    print(mensaje)

mostrar_mensaje()


````

````python
def calcular():
    resultado = 10
    return resultado

print(calcular())

resultado = calcular() 
print(resultado)
````

### Experimento de alcance

Analiza:

```python
def calcular():
    resultado = 10

calcular()
print(resultado)
```

Después predice la salida:

```python
x = 10

def cambiar():
    x = 20
    print("Dentro:", x)

cambiar()
print("Fuera:", x)
```


````python
x = 10

def cambiar():
    x = 20
    print("Dentro:", x)

cambiar()
print("Fuera:", x)
````

## 6. Docstrings

Un **docstring** es una cadena de documentación ubicada inmediatamente después de la definición de una función.

```python
def area_rectangulo(base, altura):
    """Calcula el área de un rectángulo."""
    return base * altura
```

Podemos consultar la documentación con:

```python
help(area_rectangulo)
```

o:

```python
print(area_rectangulo.__doc__)
```

Para funciones más completas conviene documentar parámetros y valor de retorno.


````python
def area_rectangulo(base, altura):
    """
    Calcula el área de un rectángulo.

    Parámetros:
        base (float): longitud de la base.
        altura (float): longitud de la altura.

    Retorna:
        float: área del rectángulo.
    """
    return base * altura

print(area_rectangulo(5, 3))
print(area_rectangulo.__doc__)
help(area_rectangulo)
````

## 7. Ejemplo integrador — Movimiento uniforme

Para velocidad constante:

$$x(t)=x_0+vt$$

La función recibe `x0`, `v` y `t`, y devuelve la posición. Luego la combinamos con un `for`, retomando contenidos anteriores.


````python
def posicion(x0, v, t):
    """
    Calcula la posición con velocidad constante.

    Parámetros:
        x0 (float): posición inicial.
        v (float): velocidad constante.
        t (float): tiempo.

    Retorna:
        float: posición en el instante t.
    """
    return x0 + v*t

tiempos = [0, 1, 2, 3, 4]

for t in tiempos:
    x = posicion(10, 5, t)
    print(f"t = {t}, x = {x}")
````
