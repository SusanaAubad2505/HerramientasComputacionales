## 2. ¿Qué es un ciclo?

Un **ciclo** o **bucle** es una estructura de control que permite ejecutar repetidamente un bloque de instrucciones.

En Python utilizaremos principalmente dos tipos:

- `while`: repite mientras una condición sea verdadera.
- `for`: recorre los elementos de una secuencia o iterable.

Conceptualmente:

```text
             ┌──────────────────┐
             │ ¿Debe continuar? │
             └────────┬─────────┘
                      │
                 Sí   │
                      ▼
             ┌──────────────────┐
             │ Ejecutar bloque  │
             └────────┬─────────┘
                      │
                      └──────────────┐
                                     │
                                     ▼
                              Volver a evaluar
```

Un ciclo necesita algún mecanismo que permita que eventualmente termine.

Si la condición nunca cambia o nunca deja de cumplirse, podemos producir un **ciclo infinito**.

## 3. Ciclo `while`

La palabra `while` significa **mientras**.

Su estructura básica es:

```python
while condicion:
    instrucciones
```

Python ejecuta las instrucciones mientras `condicion` sea `True`.

### Ejemplo

```python
contador = 1

while contador <= 5:
    print(contador)
    contador += 1
```

### ¿Qué ocurre paso a paso?

1. `contador` comienza en `1`.
2. Python evalúa `contador <= 5`.
3. La condición es verdadera, así que imprime `1`.
4. `contador += 1` aumenta el contador.
5. Se vuelve a evaluar la condición.
6. El proceso continúa hasta que `contador` vale `6`.
7. Cuando `6 <= 5` es falso, el ciclo termina.

### Tres elementos importantes

Un `while` normalmente necesita:

1. **Inicialización**
2. **Condición**
3. **Actualización**

Patrón general:

```python
variable = valor_inicial

while condicion:
    # instrucciones
    variable = actualizacion
```

````python
contador = 1

while contador <= 5:
    print(contador)
    contador += 1
````


También existen:


Estos operadores son especialmente útiles dentro de ciclos porque permiten actualizar variables de manera sencilla.

## 5. ¡Cuidado con los ciclos infinitos!

Un ciclo infinito ocurre cuando la condición de un `while` nunca llega a ser falsa.

Por ejemplo:

```python
contador = 1

while contador <= 5:
    print(contador)
```

El problema es que `contador` nunca cambia.

Por lo tanto, siempre tendremos:

```text
1 <= 5 → True
```

### Regla práctica

Cuando utilices `while`, pregúntate:

Si no puedes responder esta pregunta, probablemente tienes un problema en el algoritmo.

> **Consejo:** durante las pruebas, evita ejecutar ciclos infinitos sin una forma clara de detenerlos.

# 11. Ciclo `for`

Ahora conoceremos `for`.

Su estructura básica es:


Un `for` toma cada elemento de una secuencia o iterable y ejecuta el bloque de código.

### Ejemplo


## 16. `while` vs. `for`

Ambos ciclos permiten repetir instrucciones, pero se utilizan con enfoques diferentes.

| `while` | `for` |
|---|---|
| Repite mientras una condición sea verdadera | Recorre una secuencia o iterable |
| Es útil cuando no conocemos el número de iteraciones | Es útil cuando conocemos el recorrido |
| Depende de una condición | Depende de los elementos a recorrer |
| Debemos controlar cuidadosamente la actualización | El recorrido avanza automáticamente |

### Mismo resultado, dos soluciones

Con `while`:

```python
numero = 1

while numero <= 5:
    print(numero)
    numero += 1
```

Con `for`:

```python
for numero in range(1, 6):
    print(numero)
```

### Regla práctica

Pregúntate:

> **¿Estoy esperando a que ocurra una condición o estoy recorriendo elementos?**

Si depende de una condición → probablemente `while`.

Si estás recorriendo una secuencia → probablemente `for`.

# 18. `break`

`break` permite **terminar inmediatamente el ciclo**.

Cuando Python encuentra un `break`, abandona el ciclo y continúa con la primera instrucción que esté después de él.

### Ejemplo

```python
for numero in range(1, 11):

    if numero == 6:
        break

    print(numero)
```

El resultado es:

```text
1
2
3
4
5
```

Cuando `numero` llega a `6`, se ejecuta `break` y el ciclo termina.

````python
for numero in range(1, 11):
    if numero == 6:
        break

    print(numero)
````

## 19. `break` con `while True`

Una estrategia común consiste en utilizar:

```python
while True:
```

Esto crea un ciclo que, por sí mismo, no tiene una condición de salida.

Por eso necesitamos un `break`.

### Ejemplo

```python
while True:

    palabra = input("Escriba 'salir' para terminar: ")

    if palabra == "salir":
        break

    print("Escribió:", palabra)
```

Aquí el ciclo continúa hasta que el usuario escribe exactamente `"salir"`.

### Idea importante

`while True` no es incorrecto.

Es válido cuando existe una condición clara que utiliza `break` para terminar el ciclo.

````python
while True:
    palabra = input("Escriba 'salir' para terminar: ")

    if palabra == "salir":
        break

    print("Escribió:", palabra)
````

# 20. `continue`

`continue` **no termina el ciclo**.

En cambio, hace que Python salte el resto de la iteración actual y pase directamente a la siguiente.

### Ejemplo

```python
for numero in range(1, 11):

    if numero == 5:
        continue

    print(numero)
```

Resultado:

```text
1
2
3
4
6
7
8
9
10
```

El número `5` se omite.

### Diferencia fundamental

```text
break     → termina el ciclo
continue  → salta la iteración actual
```

Esta diferencia es fundamental.

````python
for numero in range(1, 11):
    if numero == 5:
        continue

    print(numero)
````

## 21. `pass`

`pass` significa literalmente que Python **no debe hacer nada en ese punto**.

No termina el ciclo.

No salta la iteración.

No modifica ninguna variable.

Simplemente sirve como una instrucción vacía.

### Ejemplo

```python
for numero in range(5):

    if numero == 2:
        pass

    print(numero)
```

El resultado sigue siendo:

```text
0
1
2
3
4
```

### ¿Para qué sirve?

Es especialmente útil como marcador temporal cuando sabemos que necesitamos una estructura, pero todavía no hemos escrito su implementación.

Por ejemplo:

```python
if edad >= 18:
    pass
else:
    print("Menor de edad")
```

Más adelante podemos reemplazar `pass` por el código definitivo.

````python
for numero in range(5):
    if numero == 2:
        pass

    print(numero)
````

## 22. Comparación de `break`, `continue` y `pass`

| Instrucción | ¿Termina el ciclo? | ¿Salta la iteración? | ¿Ejecuta una acción? |
|---|---:|---:|---:|
| `break` | Sí | — | Termina el ciclo |
| `continue` | No | Sí | Pasa a la siguiente iteración |
| `pass` | No | No | No hace nada |

### Para recordar

**`break` = salir**

**`continue` = continuar con la siguiente**

**`pass` = no hacer nada**

# 23. Algoritmos iterativos

Hasta ahora hemos aprendido la sintaxis de los ciclos.

Ahora vamos a utilizarla para resolver problemas.

Un **algoritmo iterativo** es un algoritmo que resuelve un problema mediante la repetición de un conjunto de pasos.

Por ejemplo:

> Sumar los números del 1 al 5.

Podemos pensar:

```text
suma = 0

sumar 1
sumar 2
sumar 3
sumar 4
sumar 5
```

En lugar de escribir cada operación manualmente, usamos un ciclo.

```python
suma = 0

for numero in range(1, 6):
    suma += numero

print(suma)
```

Resultado:

```text
15
```

Este patrón es la base de muchos algoritmos.

````python
suma = 0

for numero in range(1, 6):
    suma += numero

print("Suma:", suma)
````

# 24. Contadores

Un **contador** es una variable que permite registrar cuántas veces ocurre una determinada situación.

Normalmente comienza en cero:

```python
contador = 0
```

y aumenta cuando ocurre el evento:

```python
contador += 1
```

### Ejemplo

Contar cuántos números pares hay entre 1 y 10:

```python
contador = 0

for numero in range(1, 11):

    if numero % 2 == 0:
        contador += 1

print("Cantidad de pares:", contador)
```

El contador responde a la pregunta:

> ¿Cuántas veces se cumplió esta condición?

````python
contador = 0

for numero in range(1, 11):
    if numero % 2 == 0:
        contador += 1

print("Cantidad de pares:", contador)
````

# 25. Acumuladores

Un **acumulador** es una variable que va construyendo progresivamente un resultado.

Por ejemplo, para sumar números:

```python
suma = 0

for numero in range(1, 6):
    suma += numero
```

La variable `suma` va tomando estos valores:

```text
Inicio → 0
Después de sumar 1 → 1
Después de sumar 2 → 3
Después de sumar 3 → 6
Después de sumar 4 → 10
Después de sumar 5 → 15
```

### Diferencia

**Contador:**

```python
contador += 1
```

Cuenta eventos.

**Acumulador:**

```python
suma += valor
```

Acumula valores.

Estos dos patrones aparecen constantemente en programación.

# 26. Calcular un promedio

Para calcular un promedio necesitamos:

1. Acumular los valores.
2. Contar cuántos valores tenemos.
3. Dividir la suma entre la cantidad.

```python
suma = 0
contador = 0

for numero in range(1, 6):
    suma += numero
    contador += 1

promedio = suma / contador

print("Promedio:", promedio)
```

Aquí combinamos dos patrones:

- `suma` → acumulador.
- `contador` → contador.

### Idea algorítmica

Muchos problemas se pueden resolver identificando:

> **¿Qué necesito contar?**

y

> **¿Qué necesito acumular?**

````python
suma = 0
contador = 0

for numero in range(1, 6):
    suma += numero
    contador += 1

promedio = suma / contador

print("Promedio:", promedio)
````

# 27. Búsqueda

Otro patrón importante es buscar un elemento.

Por ejemplo, queremos determinar si el número `7` está dentro del rango del 1 al 10.

```python
numero_buscado = 7
encontrado = False

for numero in range(1, 11):

    if numero == numero_buscado:
        encontrado = True
        break

if encontrado:
    print("Número encontrado")
else:
    print("Número no encontrado")
```

Aquí utilizamos:

- `for` para recorrer.
- `if` para comparar.
- una variable booleana para guardar el resultado.
- `break` para detener la búsqueda cuando encontramos lo que buscábamos.

### ¿Por qué utilizar `break`?

Porque una vez encontrado el elemento, no necesitamos continuar recorriendo.

````python
numero_buscado = 7
encontrado = False

for numero in range(1, 11):
    if numero == numero_buscado:
        encontrado = True
        break

if encontrado:
    print("Número encontrado")
else:
    print("Número no encontrado")
````

# 31. Solución propuesta del reto

Una posible solución es:

```python
cantidad = int(input("¿Cuántas calificaciones desea ingresar? "))

suma = 0
aprobadas = 0
reprobadas = 0
mayor = None
menor = None

for i in range(cantidad):

    nota = float(input(f"Ingrese calificación {i + 1}: "))

    suma += nota

    if nota >= 3:
        aprobadas += 1
    else:
        reprobadas += 1

    if mayor is None or nota > mayor:
        mayor = nota

    if menor is None or nota < menor:
        menor = nota

promedio = suma / cantidad

print("\n--- RESULTADOS ---")
print("Promedio:", promedio)
print("Aprobadas:", aprobadas)
print("Reprobadas:", reprobadas)
print("Mayor calificación:", mayor)
print("Menor calificación:", menor)
```

### Observación importante

En este algoritmo `mayor` y `menor` comienzan como:

```python
None
```

Esto significa que todavía no tenemos un valor asignado.

En la primera iteración, la primera calificación se utiliza como referencia.

Después, cada nueva calificación se compara con el máximo y el mínimo actuales.

````python
cantidad = int(input("¿Cuántas calificaciones desea ingresar? "))

suma = 0
aprobadas = 0
reprobadas = 0
mayor = None
menor = None

for i in range(cantidad):
    nota = float(input(f"Ingrese calificación {i + 1}: "))

    suma += nota

    if nota >= 3:
        aprobadas += 1
    else:
        reprobadas += 1

    if mayor is None or nota > mayor:
        mayor = nota

    if menor is None or nota < menor:
        menor = nota

promedio = suma / cantidad

print("\n--- RESULTADOS ---")
print("Promedio:", promedio)
print("Aprobadas:", aprobadas)
print("Reprobadas:", reprobadas)
print("Mayor calificación:", mayor)
print("Menor calificación:", menor)
````

# 32. Resumen de la clase

## Ciclos

Un ciclo permite repetir instrucciones.

### `while`

Repite mientras una condición sea verdadera:

```python
while condicion:
    instrucciones
```

### `for`

Recorre una secuencia o iterable:

```python
for elemento in secuencia:
    instrucciones
```

### `range()`

Permite generar secuencias numéricas:

```python
range(fin)
range(inicio, fin)
range(inicio, fin, paso)
```

Recuerda:

> El límite final de `range()` **no se incluye**.

### Control de ciclos

```text
break     → termina el ciclo
continue  → salta a la siguiente iteración
pass      → no hace nada
```

### Patrones algorítmicos

**Contador:**

```python
contador += 1
```

**Acumulador:**

```python
suma += valor
```

**Búsqueda:**

```python
if elemento == buscado:
    encontrado = True
    break
```
