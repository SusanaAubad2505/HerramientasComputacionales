# 2. Conversión de tipos

## Conversión de tipos

```python
numero = 10

print(float(numero))
print(str(numero))
print(bool(numero))
```

Pero hay casos interesantes:

```python
print(bool(0))
print(bool(1))
print(bool(-5))
print(bool(""))
print(bool("Hola"))
```

## Conversión de cadenas

```python
numero = "25"

resultado = int(numero) + 5

print(resultado)
```

Pero:

```python
numero = "25.5"

resultado = int(numero)
```

produce un error.

¿Por qué?

Porque "25.5" no representa directamente un entero.

La solución:

```python
numero = float("25.5")
```

## Ejemplo aplicado

Supongamos que queremos calcular el promedio de tres notas.

```python
nota1 = float(input("Nota 1: "))
nota2 = float(input("Nota 2: "))
nota3 = float(input("Nota 3: "))

promedio = (nota1 + nota2 + nota3) / 3

print("Promedio:", promedio)
```

Después podemos preguntar:

¿Qué ocurriría si olvidamos float()?

```python
nota1 = input("Nota 1: ")
nota2 = input("Nota 2: ")
```

Y hacemos:

```python
promedio = (nota1 + nota2) / 2
```

Python genera un error porque estamos intentando operar cadenas como números.

```python
es_estudiante or es_profesor
```

```python
not es_mayor
```

## Ejemplo más avanzado

Una universidad ofrece un descuento si:

- el estudiante tiene promedio mayor o igual a 4.0
- o
- pertenece a un programa especial.

```python
promedio = 3.7
programa_especial = True

descuento = promedio >= 4.0 or programa_especial

print(descuento)
```

# 6. Precedencia de operadores

```python
resultado = 10 + 5 * 2
```

porque la multiplicación se evalúa antes.

Con operadores lógicos:

```python
resultado = 10 > 5 and 20 < 30
```

Python evalúa las expresiones relacionales y posteriormente el and.

Pero recomiendo enseñarles a utilizar paréntesis para hacer explícita la intención:

```python
resultado = (10 > 5) and (20 < 30)
```

Esto mejora muchísimo la legibilidad.

# 8. Comentarios

Primero:

```python
# Este programa calcula el promedio de tres notas
```

```python
nota1 = 4.5
nota2 = 3.8
nota3 = 4.2

# Calculamos el promedio
promedio = (nota1 + nota2 + nota3) / 3
```

```python
edad = 20  # Guardamos 20 en edad
```

**El código explica qué hace; los comentarios deberían explicar por qué.**

## 5. Separar entrada, procesamiento y salida

```python
# Entrada
nota1 = float(input("Nota 1: "))
nota2 = float(input("Nota 2: "))

# Procesamiento
promedio = (nota1 + nota2) / 2
aprobado = promedio >= 3.0

# Salida
print("Promedio:", promedio)
print("¿Aprobó?:", aprobado)
```
