# 1. ¿Cómo toma decisiones un programa?

Hasta ahora hemos escrito programas que ejecutan instrucciones de manera secuencial. Sin embargo, muchos problemas requieren que el programa **decida qué hacer dependiendo de una situación**.

Por ejemplo:

> Si está lloviendo, llevo paraguas. Si no está lloviendo, no llevo paraguas.

En programación podemos representar esta idea mediante una **condición**.

Una condición produce uno de dos resultados:

- `True`: la condición es verdadera.
- `False`: la condición es falsa.

Las estructuras `if`, `elif` y `else` permiten utilizar esos resultados para decidir qué instrucciones ejecutar.

````python
esta_lloviendo = True

print(esta_lloviendo)
````

# 2. Valores booleanos

El tipo de dato booleano (`bool`) solamente puede tener dos valores:

```python
True
False
```

Los booleanos aparecen constantemente cuando hacemos comparaciones.

Por ejemplo, la expresión `10 > 5` produce `True`, mientras que `10 < 5` produce `False`.

````python
print(10 > 5)
print(10 < 5)
print(type(True))
````

# 3. Operadores relacionales

Los operadores relacionales permiten comparar valores.

| Operador | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `>` | mayor que | `8 > 5` | `True` |
| `<` | menor que | `8 < 5` | `False` |
| `>=` | mayor o igual que | `8 >= 8` | `True` |
| `<=` | menor o igual que | `8 <= 5` | `False` |
| `==` | igual a | `8 == 8` | `True` |
| `!=` | diferente de | `8 != 5` | `True` |

## Atención: `=` no es lo mismo que `==`

- `=` asigna un valor a una variable.
- `==` compara dos valores.

```python
edad = 20       # asignación
edad == 20      # comparación
```

````python
edad = 20

print(edad > 18)
print(edad == 20)
print(edad != 15)
print(edad <= 17)
````

# 4. La estructura `if`

La estructura básica es:

```python
if condicion:
    instrucciones
```

Python evalúa la condición. Si el resultado es `True`, ejecuta el bloque indentado. Si es `False`, no lo ejecuta.

### Ejemplo

Queremos determinar si una persona es mayor de edad.

````python
edad = 20

if edad >= 18:
    print("Es mayor de edad")
````

### Elementos importantes

```python
if edad >= 18:
    print("Es mayor de edad")
```

1. `if` indica que se evaluará una condición.
2. `edad >= 18` es la condición.
3. `:` indica el comienzo del bloque.
4. La indentación indica qué instrucciones pertenecen al `if`.

La indentación es obligatoria en Python.

# 5. `if` + `else`

Cuando existen dos posibilidades podemos utilizar `else`.

```python
if condicion:
    instrucciones_si_es_verdadero
else:
    instrucciones_si_es_falso
```

Solo uno de los dos bloques se ejecuta.

````python
edad = int(input("Ingrese su edad: "))

if edad >= 18:
    print("Puede ingresar")
else:
    print("No puede ingresar")
````

## ¿Por qué importan los límites?

Observa la diferencia:

```python
edad > 18
```

y

```python
edad >= 18
```

Con `>` una persona de exactamente 18 años no cumple la condición. Con `>=`, sí la cumple.

# 6. `elif`: más de dos posibilidades

Cuando tenemos varias posibilidades utilizamos `elif`, abreviatura de *else if*.

```python
if condicion_1:
    ...
elif condicion_2:
    ...
elif condicion_3:
    ...
else:
    ...
```

Python evalúa las condiciones **de arriba hacia abajo**. Cuando encuentra la primera condición verdadera, ejecuta ese bloque y deja de evaluar los demás `elif`.

````python
nota = float(input("Ingrese la nota: "))

if nota >= 4.5:
    print("Excelente")
elif nota >= 4.0:
    print("Muy bueno")
elif nota >= 3.0:
    print("Aprobado")
else:
    print("Reprobado")
````

## Un error frecuente: el orden de las condiciones

Este código tiene un problema:

```python
if nota >= 3.0:
    print("Aprobado")
elif nota >= 4.5:
    print("Excelente")
```

Si `nota` vale `4.8`, la primera condición (`nota >= 3.0`) ya es verdadera. Por eso nunca se llega al `elif`.

### Regla práctica

Cuando las condiciones representan rangos, normalmente conviene comenzar por el rango más alto o más específico y avanzar hacia los rangos inferiores.

# 7. Operadores lógicos

A veces una sola comparación no es suficiente. Podemos combinar varias condiciones mediante operadores lógicos.

| Operador | Significado | Idea |
|---|---|---|
| `and` | Y | Todas las condiciones deben ser verdaderas |
| `or` | O | Al menos una condición debe ser verdadera |
| `not` | Negación | Invierte el resultado lógico |

Por ejemplo, una persona puede ingresar si tiene 18 años o más **y** tiene una entrada:

```python
edad >= 18 and tiene_entrada
```

## `and`

Con `and`, **todas** las condiciones deben ser verdaderas.

| A | B | A `and` B |
|---|---|---|
| `False` | `False` | `False` |
| `False` | `True` | `False` |
| `True` | `False` | `False` |
| `True` | `True` | `True` |

````python
edad = 20
tiene_entrada = True

if edad >= 18 and tiene_entrada:
    print("Puede ingresar")
else:
    print("No puede ingresar")
````

## `or`

Con `or`, basta con que **al menos una** condición sea verdadera.

Ejemplo: una persona puede recibir un beneficio si es estudiante **o** docente.

```python
if es_estudiante or es_docente:
    print("Tiene beneficio")
```

````python
es_estudiante = False
es_docente = True

if es_estudiante or es_docente:
    print("Tiene beneficio")
else:
    print("No tiene beneficio")
````

## `not`

`not` invierte un valor booleano:

```python
not True   # False
not False  # True
```

Puede ser útil cuando queremos preguntar si algo **no** ocurre.

````python
es_menor = False

if not es_menor:
    print("No es menor de edad")
````

# 8. Combinar condiciones

Podemos combinar operadores relacionales y lógicos.

Por ejemplo:

```python
18 <= edad <= 65
```

es equivalente a:

```python
edad >= 18 and edad <= 65
```

La primera forma es una característica muy útil de Python llamada **comparación encadenada**.

````python
edad = int(input("Ingrese su edad: "))

if 18 <= edad <= 65:
    print("Está dentro del rango")
else:
    print("Está fuera del rango")
````

# 9. `input()` y condiciones

Recuerda que `input()` devuelve siempre un texto (`str`).

Por eso, si necesitamos hacer una comparación numérica, debemos convertir el dato.

Incorrecto:

```python
edad = input("Edad: ")
if edad >= 18:
    print("Mayor de edad")
```

Correcto:

```python
edad = int(input("Edad: "))
if edad >= 18:
    print("Mayor de edad")
```

Para texto sí podemos comparar cadenas directamente:

````python
respuesta = input("¿Es estudiante? (si/no): ")

if respuesta == "si":
    print("Tiene descuento")
else:
    print("No tiene descuento")
````

# 12. Ejercicio integrador — Sistema de cobro de parqueadero

Un parqueadero cobra:

- Menos de 2 horas → `$5.000`
- Entre 2 y 5 horas → `$8.000`
- Más de 5 horas → `$12.000`

Además, los estudiantes reciben un descuento del 20%.

### Requisitos

El programa debe:

1. Solicitar el número de horas.
2. Solicitar si el cliente es estudiante.
3. Determinar el precio según las horas.
4. Aplicar el descuento si corresponde.
5. Mostrar el total a pagar.

### Pregunta para discutir

> ¿Es necesario repetir la condición de estudiante dentro de cada `if` del precio?

Intenta resolver el problema separando las decisiones en pasos.

````python
# Solución propuesta

horas = int(input("Horas de parqueadero: "))
estudiante = input("¿Es estudiante? (si/no): ")

if horas < 2:
    precio = 5000
elif horas <= 5:
    precio = 8000
else:
    precio = 12000

if estudiante == "si":
    precio = precio * 0.80

print("Total a pagar:", precio)
````

# 13. Errores frecuentes

## Error 1 — Usar `=` en lugar de `==`

```python
if edad = 18:
```

❌ Incorrecto.

```python
if edad == 18:
```

✅ Correcto.

## Error 2 — Olvidar `:`

```python
if edad >= 18
```

❌ Incorrecto.

```python
if edad >= 18:
```

✅ Correcto.

## Error 3 — Mala indentación

```python
if edad >= 18:
print("Mayor")
```

❌ Incorrecto.

```python
if edad >= 18:
    print("Mayor")
```

✅ Correcto.

## Error 4 — Orden incorrecto de condiciones

Si una condición general aparece antes que una condición más específica, puede impedir que el programa llegue al `elif` esperado.

## Error 5 — Olvidar la conversión de `input()`

Para cálculos y comparaciones numéricas, normalmente debemos convertir:

```python
edad = int(input("Edad: "))
nota = float(input("Nota: "))
```
